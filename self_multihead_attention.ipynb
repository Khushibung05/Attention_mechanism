###Self Attention & Multi-head attention

```python
import numpy as np
import tensorflow as tf

from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input,
    LSTM,
    Embedding,
    Dense,
    MultiHeadAttention,
    GlobalAveragePooling1D,
    Concatenate
)

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

####Dataset

```python
sentences = [
"The animal didn't cross the street because it was tired",
"The trophy doesn't fit in the suitcase because it is too large",
"The book was placed on the table because it was heavy",
"The laptop was returned because it was damaged",
"The car stopped because it ran out of fuel"
]

labels = [0,1,2,3,4]
```

####Tokenization

```python
tokenizer=Tokenizer()
tokenizer.fit_on_texts(sentences)
sequences=tokenizer.texts_to_sequences(sentences)
x=pad_sequences(sequences,padding='post')
y=np.array(labels)
print(x.shape)
```

####Input Layer

```python
inputs=Input(shape=(x.shape[1],))
```

####Embedding Layer

```python
embedding=Embedding(
    input_dim=len(tokenizer.word_index)+1,
    output_dim=32,
)(inputs)
```

####Multihead self Attention

```python
attention_layer=MultiHeadAttention(
    num_heads=2,
    key_dim=32,
)

attention_output,attention_scores=attention_layer(
    embedding,
    embedding,
    return_attention_scores=True
)
```

####Pooling

pooling converts: (batch_size,sequnce_length,embedding_dim)

to:(batch_size,embedding_dim)

```python

pooled=GlobalAveragePooling1D()(attention_output)
```

####Dense layers

```python
dense=Dense(64,activation='relu')(pooled)
outputs=Dense(5,activation='softmax')(dense)
```

####Build model

```python
model=Model(inputs,outputs)
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
model.summary()
```

####Train model

```python
model.fit(
    x,
    y,
    epochs=50,verbose=1
)
```

####Prediction

```python
test_sentence = [
    "The animal did not cross the street because it was tired"
]

test_seq = tokenizer.texts_to_sequences(test_sentence)

test_seq = pad_sequences(
    test_seq,
    maxlen=x.shape[1],
    padding="post"
)

prediction = model.predict(test_seq, verbose=0)

predicted_class = np.argmax(prediction)

print("\nPredicted Class:", predicted_class)
```

####Label Mapping

```python
label_mapping={
    0:'animal',
    1:'trophy',
    2:'book',
    3:'laptop',
    4:'car'
}
print(label_mapping[predicted_class])
```

```python
# ATTENTION SCORE MODEL
# =====================================================

attention_model = Model(
    inputs=inputs,
    outputs=attention_scores
)

scores = attention_model.predict(test_seq)

print("\nAttention Score Shape:")
print(scores.shape)

# shape:
# (batch_size, num_heads, seq_len, seq_len)

tokens = test_sentence[0].lower().split()

print("\nTokens:")
print(tokens)
```

```python
# =====================================================
# DISPLAY HEAD-1 ATTENTION MATRIX
# =====================================================

head1 = scores[0, 0]

print("\nHead 1 Attention Matrix:\n")

for i in range(len(tokens)):
    print(f"\nToken: {tokens[i]}")

    for j in range(len(tokens)):
        print(
            f"  -> {tokens[j]:10s}: {head1[i][j]:.4f}"
        )
```

```python
# =====================================================
# DISPLAY HEAD-2 ATTENTION MATRIX
# =====================================================

head2 = scores[0, 1]

print("\n\nHead 2 Attention Matrix:\n")

for i in range(len(tokens)):
    print(f"\nToken: {tokens[i]}")

    for j in range(len(tokens)):
        print(
            f"  -> {tokens[j]:10s}: {head2[i][j]:.4f}"
        )
```
